//coefficicients <== []

//notes <== []

//pour chaque ressource de la liste ressources :

////ajouter à la liste coefficients; int(la demande du coefficient de telle {ressource} à l'UE1)

////ajouter à la liste note; float(la demande de  la note obtenu pour telle {ressource})

//somme = 0

//somme_coefs = 0

//pour k allant jusqu'à la longueur de la liste coefficients:

////ajouter à la variable somme la valeur : coefficients[k] * notes[k]

////ajouter à la variable somme_coefs la valeur : coefficients[k]

//moyenne <== somme / somme_coefs

//afficher moyenne


```python
def calcul_moyenne_UE1():
    ressources = [f"R{i}" for i in range(101,116)] + [f"SAE{k}" for k in range(11,17)]
    coefficients = []
    notes = []
    for ressource in ressources:
        coefficients.append(int(input(f"Renseigner le coefficient de la ressource {ressource}: ")))
        notes.append(float(input(f"Renseigner la note que vous avez obtenue en {ressource}" )))
    somme = 0
    somme_coefs = 0
    for k in range(len(coefficients)):
        somme += (coefficients[k] * notes[k])
        somme_coefs += coefficients[k]
    moyenne = somme / somme_coefs
    print(f" Vous avez une moyenne de {moyenne} à l'UE1 !")
calcul_moyenne_UE1()
```
