// note_R101 <== float(demander à l'utilisateur la note obtenu en R101)

// coef_R102 <== int(demander le coefficient de la R102 (UE1))

// somme <== (note_R101* coef R101 + note_R102 * coef_R102) / (coef_R101 + coef R102)

// affichier somme

appeler la fonction calcul_moyenne_UE1

```python
def calcul_moyenne_UE1():
    note_R101 = float(input("Note obtenue en R101 : "))
    coef_R101 = float(input("Coefficient de R101 (UE1) : "))
    note_R102 = float(input("Note obtenue en R102 : "))
    coef_R102 = float(input("Coefficient de R102 (UE1) : "))
    moyenne = (note_R101 * coef_R101 + note_R102 * coef_R102) / (coef_R101 + coef_R102)
    print("La moyenne de l'UE1 est :", moyenne)
calcul_moyenne_UE1()
```
