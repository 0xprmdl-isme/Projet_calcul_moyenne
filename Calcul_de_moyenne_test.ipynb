```python
def calcul_moyenne_UE1():
    ressources = [f"R{i}" for i in range(101,116)] + [f"SAE{k}" for k in range(11,17)]
    coefficients = []
    somme_coefs = 0
    notes = []
    for ressource in ressources:
        coef = int(input(f"Renseigner le coefficient de la ressource {ressource}: "))
        coefficients.append(coef)
        somme_coefs += coef
        notes.append(float(input(f"Renseigner la note que vous avez obtenue en {ressource}" )))
    dictionnaire_coefs = {ressource: coef for ressource,coef in zip(ressources,coefficients)}
    dictionnaire_notes = {ressource: note for ressource,note in zip(ressources,notes)}
    somme =0
    for cle,value in dictionnaire_coefs.items():
        somme+= value*dictionnaire_notes[cle]
    moyenne = somme / somme_coefs
    print(f"Vous avez une moyenne de {moyenne} à l'UE1 !")
calcul_moyenne_UE1()
    
```
